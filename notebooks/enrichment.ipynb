```python
import polars as pl
#read from parquet
df = pl.read_parquet(r"C:\projects\interactome\data\processed_networks\BioPlex.parquet")
df.columns
```

```python
import networkx as nx
#use first two columns of df to create a graph and third weight
G = nx.from_pandas_edgelist(
    df,
    source=df.columns[0],
    target=df.columns[1],
    edge_attr=df.columns[2],
    create_using=nx.Graph(),
)
G
```

```python
from IPython.display import display
import matplotlib.pyplot as plt
import networkx as nx
import polars as pl

# Graph summary
n_nodes = G.number_of_nodes()
n_edges = G.number_of_edges()
components = list(nx.connected_components(G))
largest_component = max(components, key=len) if components else set()

summary = pl.DataFrame({
    "metric": [
        "nodes", "edges", "density", "connected components",
        "largest component size", "mean degree", "average clustering",
    ],
    "value": [
        n_nodes,
        n_edges,
        nx.density(G) if n_nodes > 1 else 0.0,
        len(components),
        len(largest_component),
        sum(dict(G.degree()).values()) / n_nodes if n_nodes else 0.0,
        nx.average_clustering(G) if n_nodes else 0.0,
    ],
})
display(summary)

# Degree distribution
```

```python
import matplotlib.pyplot as plt

degrees = [degree for _, degree in G.degree()]
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(degrees, bins="auto", edgecolor="white")
axes[0].set(title="Degree distribution", xlabel="Degree", ylabel="Number of proteins")

# Plot the largest component (layout can be slow for large graphs)
if largest_component:
    H = G.subgraph(largest_component)
    pos = nx.spring_layout(H, seed=42)
    nx.draw_networkx(
        H, pos=pos, ax=axes[1], with_labels=False,
        node_size=20, edge_color="lightgray", width=0.5,
    )
axes[1].set_title(f"Largest component ({len(largest_component)} proteins)")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
